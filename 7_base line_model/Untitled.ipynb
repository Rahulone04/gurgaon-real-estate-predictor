```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('gurgaon_properties_post_feature_selection.csv')
```

```python
df.head()
```

```python
# For applying Linear Regression we have to do one hot encode
#one hot encode--> sector,balcony,age progression, furnishing type, Luxuary category, floor category
```

```python
X=df.drop(columns=['price'])
y=(df['price'])
```

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
columns_to_encode=['sector','balcony','agePossession','funishing_type','luxury_category', 'floor_category']
```

```python
# Applying the Logp1 transformation to the target variable(beacuse price column is right skewed we are doing this to convert it into center skewed)
y_transformed= np.log1p(y)
```

```python
# Creating a column transformer for preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), ['property_type', 'bedRoom', 'bathroom', 'built_up_area', 'servant room', 'store room']),
        ('cat', OneHotEncoder(drop='first'), columns_to_encode)
    ], 
    remainder='passthrough'
)
```

```python
# creating a pipeline

pipeline=Pipeline([
    ('preprocessor',preprocessor),
    ('regressor',LinearRegression())
])
```

```python

# K-fold cross-validation
kfold = KFold(n_splits=10, shuffle=True, random_state=42)
scores = cross_val_score(pipeline, X, y_transformed, cv=kfold, scoring='r2')
```
