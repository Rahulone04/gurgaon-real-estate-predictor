```python
# Column to drop jiska koi use nahi hai analysis mai ->property_name, link , property id but we are not going to drop the property_name wala column as it contains sector info which can be used further

df.drop(columns=['link','property_id'],inplace=True)
```

```python
# Society column cleaning
# First check how many unique categories are there in society column
df['society'].value_counts()
```

```python
# Now we will remove the society names which are same but has difference due to symbol and ranting it makes a new unique category
import re
df['society']=df['society'].apply(lambda name: re.sub(r'\d+(\.\d+)?\s?[★*?]?','',str(name)).strip()).str.lower()
```

```python
# IN price column we change the lac unit into crore and also if the price is not given and in place of price price on request is given then we will drop that column
df=df[df['price'] !='Price on Request']
```

```python
# Now we will make a fuction which take lac price as a input and convert it into crore wala unit

def treat_price(x):
    if type(x) == float:
        return x
    else:
        if x[1]=='Lac':
            return round(float(x[0])/100,2)
        else:
            return round(float(x[0]),2)
```

```python
df['price']=df['price'].str.split(' ').apply(treat_price)
```

```python
df.sample(10)
```

```python
# In area column firly i want to change the name of the column 

df.rename(columns={'area':'price_per_sqft'},inplace=True)
```

```python
df.head()
```

```python
# Now we will modify the price_per_sqft wala column 

df['price_per_sqft']=df['price_per_sqft'].str.split('/').str.get(0).str.replace('?','').str.replace(',','').str.strip().astype('float')
```
